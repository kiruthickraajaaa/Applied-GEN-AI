Dictionary Operations

```python
fruit_inventory = {"apple": 50,"banana": 120,"cherry": 30,"berry":57,"mango":34}
print(fruit_inventory["apple"])
print(fruit_inventory["mango"])
```

```python
#adding
fruit_inventory["orange"] = 80
print(fruit_inventory)
#updating
fruit_inventory.update({"banana": 110, "mango": 45})
print(fruit_inventory)
#deleting
del fruit_inventory["cherry"]
print(fruit_inventory)
```

```python
for fruit in fruit_inventory:
  print(fruit)
```

Set Operations

```python
basket_a = {"apple", "banana", "cherry"}
basket_b = {"cherry", "dates", "mango", "banana"}
```

```python
#union
print(basket_a.union(basket_b))
#intersection
print(basket_a.intersection(basket_b))
#difference
print(basket_a.difference(basket_b))
```

```python
#union
print(basket_a | basket_b)
#intersection
print(basket_a & basket_b)
#difference
print(basket_a - basket_b)
```

Conditional Statements

```python
score=int(input("Enter Your Score : "))
if score>=90 and score<=100:
  print("A Grade")
elif score>=80 and score<90:
  print("B Grade")
elif score>=70 and score<80:
    print("C Grade")
elif score>=60 and score<70:
    print("D Grade")
elif score>=50 and score <60:
    print("E Grade")
elif score<50 and score>=0:
    print("Fail")
else:
  print("Invalid Input")
```

program to count the frequency of elements in a list.

```python
words = [
    "apple", "banana", "apple", "orange",
    "banana", "apple", "grape", "orange", "apple"
]
```

```python
words = ["apple", "banana", "apple", "orange", "banana", "apple"]

frequency = {}

for item in set(words):
    frequency[item] = words.count(item)

print(frequency)
```
