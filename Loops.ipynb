***While Looop***

```python
#print from 1 to 20
#initialization,Condition,increament or decrement

c=1
while c<=20:
  print(c)
  c+=1
```

```python
#print 20 to 1
c=20
while c>=1:
  print(c)
  c-=1
```

**For Loop with conditional Statement**

```python
for i in range(1,11,1):

   if(i%2==0):
    print(i,"Even")
   else:
    print(i,"odd")
```

```python
#print 1 to 10 without 5

for i in range(1, 11):
    if i != 5:
        print(i)
```

```python
for i in range(1, 11):
  if i==5:
    pass
  else:
    print(i)
```

```python
for i in range(1, 11):
  if i==5:
    continue
  else:
    print(i)
```

```python
#print 5,15,20,24,30
for i in range(5,31,5):
  if(i==10):
    continue
  else:
    print(i)
```

```python
#print 5,15,20,24,30
for i in range(5,31,5):
  if(i==10):
    break
  else:
    print(i)
```
